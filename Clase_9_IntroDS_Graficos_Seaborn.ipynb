```
import pandas as pd
import matplotlib.pyplot as plt #Librería para crear graficos
import numpy as np              #Libreria para generar datos numéricos, aqui la usamos para generar datos aleatorios
import seaborn as sns
```

```
df = pd.read_csv('https://raw.githubusercontent.com/Dannap7337/Pato/main/Base_limpiaclase.csv')
df = df.drop(columns=['Unnamed: 0'])
df
```

```
df['Estadia'].unique()
```

```
df.columns
```

```
#Creación de Grafico de Distribucion
sns.kdeplot(df['Balance'], fill=True,)   #Fill rellana el interior
plt.show()
```

```
#Creación de Grafico de Distribucion
sns.kdeplot(df['Balance'], fill=False)   #Fill = False sin relleno
plt.show()
```

```
#Creación de Grafico de Distribucion
plt.figure(figsize=(10,4))                            #Medidas del gráfico
sns.kdeplot(df['Balance'], fill=True, color="red")    #Fill = False sin relleno
plt.title('Gráfica de densidad de balance')           #Pone título al gráfico
plt.show()
```

```
plt.figure(figsize=(10,4))                            #Medidas del gráfico
sns.kdeplot(df['Salario'], fill=True, color="green")    #Fill = False sin relleno
plt.title('Gráfica de densidad de salario')           #Pone título al gráfico
plt.show()
```

```
plt.figure(figsize=(10,4))                            #Medidas del gráfico
sns.kdeplot(df['CreditScore'], fill=True, color="brown")    #Fill = False sin relleno
plt.title('Gráfica de densidad de CreditScore')           #Pone título al gráfico
plt.show()
```

```
plt.figure(figsize=(10,4))                            #Medidas del gráfico
sns.kdeplot(df['Edad'], fill=True, color="yellow")    #Fill = False sin relleno
plt.title('Gráfica de densidad de Edad')           #Pone título al gráfico
plt.show()
```

```
# Crear figura y subplots (2 filas, 2 columnas)
fig, axes = plt.subplots(2, 2, figsize=(6, 4))

# KDE para Balance
sns.kdeplot(df['Balance'], fill=True, ax=axes[0, 0])
axes[0, 0].set_title('Distribución de Balance')

# KDE para Salario
sns.kdeplot(df['Salario'], fill=True, ax=axes[0, 1], color = "red")
axes[0, 1].set_title('Distribución de Salario')

# KDE para CreditScore
sns.kdeplot(df['CreditScore'], fill=True, ax=axes[1, 0], color = "yellow")
axes[1, 0].set_title('Distribución de CreditScore')

# KDE para Edad
sns.kdeplot(df['Edad'], fill=True, ax=axes[1, 1], color = "orange")
axes[1, 1].set_title('Distribución de Edad')

# Ajustar el layout para evitar superposición de títulos y etiquetas
plt.tight_layout()

# Mostrar gráfico
plt.show()
```

# **Gráfico de Pastel**

```
#Contar la cantidad de clientes por género
gender_counts = df['Genero'].value_counts()

#Gráfico de Pastel
plt.figure(figsize=(6,6))
plt.pie(gender_counts, labels=gender_counts.index, autopct='%1.1f%%', startangle=90,
        colors=['skyblue', 'lightgreen'])
plt.title('Distribución por Género')
plt.show()
```

```
# Contar la cantidad de clientes por país
pais_counts = df['Pais'].value_counts()

# Gráfico de pastel
plt.figure(figsize=(6,6))
plt.pie(pais_counts, labels=pais_counts.index, autopct='%1.1f%%', startangle=90, colors=['lightcoral', 'skyblue', 'lightgreen', 'yellow'])
plt.title('Distribución por País')
plt.show()
```

```
# Contar la cantidad de clientes por tipo de tarjeta
card_type_counts = df['Tipo_tarjeta'].value_counts()

# Gráfico de pastel
plt.figure(figsize=(5,5))
plt.pie(card_type_counts, labels=card_type_counts.index, autopct='%1.1f%%', startangle=90, colors=['purple', 'orange', 'green', 'blue'])
plt.title('Distribución por Tipo de Tarjeta')
plt.show()
```

# **Gráfico de Violin**

```
# Gráfico de violín para CreditScore por Tipo de Tarjeta
sns.violinplot(x='CreditScore', data=df)

# Mostrar gráfico
plt.title('Distribución de CreditScore')
plt.show()
```
